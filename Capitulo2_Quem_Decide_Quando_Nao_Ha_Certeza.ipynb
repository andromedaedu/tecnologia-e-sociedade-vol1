## A árvore de decisão (Etapas 1, 2 e 3)

O algoritmo abaixo recebe cinco informações sobre um cenário e recomenda **AGIR**, **AGUARDAR** ou **ABORTAR**.

- **Etapa 1 — A configuração inicial:** execute a célula sem mudar nada.
- **Etapa 2 — Reproduzir a decisão real:** mude só a **PARTE 2** até o sistema recomendar **AGIR**. Anote no livro o que você mudou.
- **Etapa 3 — Um ponto a menos:** volte à configuração que você encontrou na Etapa 2 e mude **um único dado** da PARTE 1 em 1 ponto. A decisão mudou?

```python
# ===== PARTE 1: O CENÁRIO (dados fictícios, de 0 a 10; não mude) =====
confiabilidade      = 5   # quão confiáveis são os dados sobre o alvo
risco_civis         = 3   # risco estimado de atingir civis
pressao_tempo       = 9   # urgência: o alvo pode sumir a qualquer momento
importancia         = 8   # importância estratégica atribuída ao alvo
dependencia_automacao = 8 # quanto a identificação dependeu de sistemas automáticos

# ===== PARTE 2: AS REGRAS DO ALGORITMO (MUDE AQUI) =====
LIMITE_RISCO     = 4    # risco para civis a partir do qual o sistema aborta
LIMITE_CONFIANCA = 6    # confiabilidade mínima para considerar agir
LIMIAR_ACAO      = 7    # índice mínimo para recomendar ação

PESO_CONFIABILIDADE = 0.4
PESO_PRESSAO        = 0.1
PESO_IMPORTANCIA    = 0.2
PESO_VERIFICACAO    = 0.3

# ===== PARTE 3: A ÁRVORE DE DECISÃO =====
verificacao_humana = 10 - dependencia_automacao

indice = (confiabilidade * PESO_CONFIABILIDADE
          + pressao_tempo * PESO_PRESSAO
          + importancia * PESO_IMPORTANCIA
          + verificacao_humana * PESO_VERIFICACAO)

if risco_civis >= LIMITE_RISCO:
    decisao = "ABORTAR"
    motivo = "o risco para civis passou do limite"
elif confiabilidade < LIMITE_CONFIANCA:
    decisao = "AGUARDAR"
    motivo = "os dados não são confiáveis o suficiente"
elif indice >= LIMIAR_ACAO:
    decisao = "AGIR"
    motivo = "o índice alcançou o limiar de ação"
else:
    decisao = "AGUARDAR"
    motivo = "o índice ficou abaixo do limiar de ação"

# ===== PARTE 4: O RESULTADO =====
print("Soma dos pesos:", round(PESO_CONFIABILIDADE + PESO_PRESSAO + PESO_IMPORTANCIA + PESO_VERIFICACAO, 2))
print("Índice calculado:", round(indice, 2), "(limiar de ação:", LIMIAR_ACAO, ")")
print("RECOMENDAÇÃO DO SISTEMA:", decisao)
print("Motivo:", motivo)
```

## Os 100 casos (Etapa 4)

Um sistema automático deu um **sinal de ameaça**, de 0 a 10, para 100 pessoas fictícias. Só nós sabemos a verdade: **10 eram ameaças reais e 90 eram inocentes**.

O sistema recomenda agir contra quem tiver sinal igual ou maior que o **LIMIAR**. Mude o limiar e execute de novo. Anote os resultados na tabela do livro.

```python
# ===== OS 100 CASOS FICTÍCIOS (não mude) =====
# Cada número é o "sinal de ameaça" (0 a 10) que um sistema automático deu a uma pessoa.
# Só o livro sabe a verdade: 10 pessoas eram ameaças reais; 90 eram inocentes.
sinais_ameacas   = [4, 5, 6, 6, 7, 7, 8, 8, 9, 9]
sinais_inocentes = ([0] * 8 + [1] * 12 + [2] * 16 + [3] * 16 + [4] * 13
                    + [5] * 10 + [6] * 7 + [7] * 4 + [8] * 3 + [9] * 1)

# ===== O LIMIAR (MUDE AQUI) =====
LIMIAR = 7   # o sistema recomenda agir contra quem tiver sinal igual ou maior que este

# ===== A CONTAGEM =====
ameacas_detectadas = 0
for s in sinais_ameacas:
    if s >= LIMIAR:
        ameacas_detectadas = ameacas_detectadas + 1

inocentes_atingidos = 0
for s in sinais_inocentes:
    if s >= LIMIAR:
        inocentes_atingidos = inocentes_atingidos + 1

ameacas_perdidas = len(sinais_ameacas) - ameacas_detectadas
total_acoes = ameacas_detectadas + inocentes_atingidos

print("Limiar:", LIMIAR)
print("Ameaças detectadas:", ameacas_detectadas, "de", len(sinais_ameacas))
print("Ameaças que escaparam (falsos negativos):", ameacas_perdidas)
print("Inocentes apontados como ameaça (falsos positivos):", inocentes_atingidos, "de", len(sinais_inocentes))
if total_acoes > 0:
    print("De cada 100 ações recomendadas,", round(100 * inocentes_atingidos / total_acoes), "atingiriam inocentes.")
```

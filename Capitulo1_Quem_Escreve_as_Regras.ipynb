## O algoritmo da bolsa (Etapas 1, 2 e 3)

Uma escola vai dar **uma** bolsa de intercâmbio no Canadá. Seis estudantes se candidataram. O algoritmo abaixo calcula a pontuação de cada um e decide quem ganha.

- **Etapa 1 — Os pesos oficiais:** execute a célula sem mudar nada.
- **Etapa 2 — Pesos iguais:** mude os quatro pesos para `0.25` e execute de novo.
- **Etapa 3 — O desafio do engenheiro:** sempre com a soma dos pesos igual a 1, encontre pesos que façam cada um dos seis candidatos ganhar. Anote os pesos na tabela do livro.

Só mude a **PARTE 2**.

```python
# ===== PARTE 1: DADOS DOS CANDIDATOS (não mude) =====
candidatos = [
    {"nome": "Ana",    "media": 8.5, "ingles": 9.5, "entrevista": 6.0, "extras": 4.0},
    {"nome": "Bruno",  "media": 7.0, "ingles": 6.0, "entrevista": 9.0, "extras": 8.0},
    {"nome": "Carla",  "media": 9.5, "ingles": 7.0, "entrevista": 7.0, "extras": 5.0},
    {"nome": "Diego",  "media": 6.5, "ingles": 8.5, "entrevista": 8.0, "extras": 6.0},
    {"nome": "Elisa",  "media": 8.0, "ingles": 5.5, "entrevista": 8.5, "extras": 9.5},
    {"nome": "Felipe", "media": 7.5, "ingles": 9.0, "entrevista": 8.5, "extras": 2.5},
]

# ===== PARTE 2: PESOS DO ALGORITMO (MUDE AQUI) =====
PESO_MEDIA      = 0.4
PESO_INGLES     = 0.4
PESO_ENTREVISTA = 0.1
PESO_EXTRAS     = 0.1

# ===== PARTE 3: A REGRA =====
def pontuacao(c):
    return (c["media"] * PESO_MEDIA
            + c["ingles"] * PESO_INGLES
            + c["entrevista"] * PESO_ENTREVISTA
            + c["extras"] * PESO_EXTRAS)

# ===== PARTE 4: ORDENAÇÃO E RESULTADO =====
ranking = sorted(candidatos, key=pontuacao, reverse=True)

print("Soma dos pesos:", round(PESO_MEDIA + PESO_INGLES + PESO_ENTREVISTA + PESO_EXTRAS, 2))
posicao = 1
for c in ranking:
    print(posicao, "-", c["nome"], "->", round(pontuacao(c), 2), "pontos")
    posicao = posicao + 1
print("BOLSA PARA:", ranking[0]["nome"])
```

## Etapa 5 — A mesma variável, dois usos opostos

A direção decidiu incluir a renda no cálculo, e os professores se dividiram:

- **Grupo Inclusão:** quem tem menos renda ganha pontos extras, porque teve menos oportunidades.
- **Grupo Custo:** quem tem mais renda ganha pontos extras, porque tem mais chance de se manter no Canadá.

Com os pesos oficiais, execute a célula nos três modos: `"nenhum"`, `"inclusao"` e `"custo"`. Só mude as linhas marcadas com **MUDE AQUI**.

```python
# ===== DADOS (não mude) =====
candidatos = [
    {"nome": "Ana",    "media": 8.5, "ingles": 9.5, "entrevista": 6.0, "extras": 4.0, "renda": 5},
    {"nome": "Bruno",  "media": 7.0, "ingles": 6.0, "entrevista": 9.0, "extras": 8.0, "renda": 2},
    {"nome": "Carla",  "media": 9.5, "ingles": 7.0, "entrevista": 7.0, "extras": 5.0, "renda": 3},
    {"nome": "Diego",  "media": 6.5, "ingles": 8.5, "entrevista": 8.0, "extras": 6.0, "renda": 4},
    {"nome": "Elisa",  "media": 8.0, "ingles": 5.5, "entrevista": 8.5, "extras": 9.5, "renda": 1},
    {"nome": "Felipe", "media": 7.5, "ingles": 9.0, "entrevista": 8.5, "extras": 2.5, "renda": 5},
]

# ===== PESOS OFICIAIS =====
PESO_MEDIA      = 0.4
PESO_INGLES     = 0.4
PESO_ENTREVISTA = 0.1
PESO_EXTRAS     = 0.1

# ===== PARÂMETROS NOVOS =====
MODO_RENDA = "nenhum"     # MUDE AQUI: "nenhum", "inclusao" ou "custo"
PESO_RENDA = 0.5          # MUDE AQUI

def bonus_renda(c):
    if MODO_RENDA == "inclusao":
        return (5 - c["renda"]) * PESO_RENDA
    elif MODO_RENDA == "custo":
        return c["renda"] * PESO_RENDA
    else:
        return 0

def pontuacao(c):
    return (c["media"] * PESO_MEDIA
            + c["ingles"] * PESO_INGLES
            + c["entrevista"] * PESO_ENTREVISTA
            + c["extras"] * PESO_EXTRAS
            + bonus_renda(c))

ranking = sorted(candidatos, key=pontuacao, reverse=True)

print("Modo da renda:", MODO_RENDA)
posicao = 1
for c in ranking:
    print(posicao, "-", c["nome"], "->", round(pontuacao(c), 2), "pontos")
    posicao = posicao + 1
print("BOLSA PARA:", ranking[0]["nome"])
```
